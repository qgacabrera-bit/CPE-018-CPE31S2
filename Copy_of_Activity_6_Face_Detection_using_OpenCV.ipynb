### Performing face detection on still image

The first and most basic way to perform face detection is to load an image and detect faces in it. To make the result visually meaningful, we will draw rectangles around faces on the original image.

**Before implementing the code below**, check the contents of the `cv2.CascadeClassifier()` function. Provide an explanation of the function, its parameters before running the code below.

```python
# Make sure that for this activity, you have downloaded the
# file indicated below from the resource linked in the instructional materials
# in the module.

import cv2
from google.colab.patches import cv2_imshow

picPath = '/content/drive/MyDrive/3rd Year Subs/Emtech 1/Act6/breakingbad.png'
haarPath = '/content/drive/MyDrive/3rd Year Subs/Emtech 1/Act6/haarcascade_frontalface_default.xml'

def faceDetect(picPath):
  face_cascade = cv2.CascadeClassifier(haarPath)

  img = cv2.imread(picPath)
  gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
  faces = face_cascade.detectMultiScale(gray, 1.2, 5)

  for (x, y, w, h) in faces:
    img = cv2.rectangle(img, (x, y), (x+w, y+h), (255,0,0), 2)

  cv2_imshow(img)
```

```python
faceDetect(picPath)
```

```python
import cv2
from google.colab.patches import cv2_imshow

def detect():
  face_cascade = cv2.CascadeClassifier('/content/drive/MyDrive/3rd Year Subs/Emtech 1/Act6/haarcascade_frontalface_default.xml')
  eye_cascade = cv2.CascadeClassifier('/content/drive/MyDrive/3rd Year Subs/Emtech 1/Act6/haarcascade_eye.xml')


  video_path = '/content/drive/MyDrive/3rd Year Subs/Emtech 1/Act6/Badvideo.mp4'
  camera = cv2.VideoCapture(video_path)


  while (True):
    ret, frame = camera.read()

    # Check if frame was successfully read
    if not ret:
      print("End of video stream or failed to read frame.")
      break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    faces = face_cascade.detectMultiScale(gray, 1.1, 5)

    for (x,y,w,h) in faces:
      img = cv2.rectangle(frame,(x,y),(x+w,y+h),(255,0,0),2)
      roi_gray = gray[y:y+h, x:x+w]
      # Ensure roi_gray is not empty before proceeding to eye detection
      if roi_gray.size == 0:
          continue
      eyes = eye_cascade.detectMultiScale(roi_gray, 1.03, 5, 0, (40,40))

      for (ex,ey,ew,eh) in eyes:
        # Corrected coordinates for drawing eyes on the original frame
        cv2.rectangle(img,(x+ex,y+ey),(x+ex+ew,y+ey+eh),
        (0,255,0),2)

    cv2_imshow(frame)

  camera.release()
  print("Video processing finished.")

if __name__ == "__main__":
  detect()
```

### Performing face and smile detection on video

To detect smiles, we will use an additional Haar cascade classifier for smiles. The logic will be similar to eye detection, where smiles are detected within the region of interest of a detected face.

```python
import cv2
from google.colab.patches import cv2_imshow

def detect_face_and_smile():
  face_cascade = cv2.CascadeClassifier('/content/drive/MyDrive/3rd Year Subs/Emtech 1/Act6/haarcascade_frontalface_default.xml')
  smile_cascade = cv2.CascadeClassifier('/content/drive/MyDrive/3rd Year Subs/Emtech 1/Act6/haarcascade_smile.xml')
  eye_cascade = cv2.CascadeClassifier('/content/drive/MyDrive/3rd Year Subs/Emtech 1/Act6/haarcascade_eye.xml')

  video_path = '/content/drive/MyDrive/3rd Year Subs/Emtech 1/Act6/raw.gif'
  camera = cv2.VideoCapture(video_path)

  while (True):
    ret, frame = camera.read()

    if not ret:
      print("End of video stream or failed to read frame.")
      break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    faces = face_cascade.detectMultiScale(gray, 1.1, 5)

    for (x,y,w,h) in faces:
      # Draw rectangle around face (Blue)
      img = cv2.rectangle(frame,(x,y),(x+w,y+h),(255,0,0),2)
      roi_gray = gray[y:y+h, x:x+w]
      roi_color = frame[y:y+h, x:x+w]

      if roi_gray.size == 0: # Check if ROI is empty
          continue
      eyes = eye_cascade.detectMultiScale(roi_gray, 1.03, 5, 0, (40,40))

      smiles = smile_cascade.detectMultiScale(
          roi_gray,
          scaleFactor=1.1,
          minNeighbors=22,
          minSize=(1w0, 10),
          flags=cv2.CASCADE_SCALE_IMAGE
      )
      for (ex,ey,ew,eh) in eyes:
        cv2.rectangle(img,(x+ex,y+ey),(x+ex+ew,y+ey+eh),
        (0,255,0),2)


      for (sx, sy, sw, sh) in smiles:
          cv2.rectangle(roi_color, (sx, sy), (sx+sw, sy+sh), (0, 0, 255), 2)

    cv2_imshow(frame)

    # Press 'q' to quit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

  camera.release()
  cv2.destroyAllWindows()
  print("Video processing finished.")

if __name__ == "__main__":
  detect_face_and_smile()
```
